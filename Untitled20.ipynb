```python
import pandas as pd
emotions.set_format(type ="pandas")
df = emotions["train"][:]
df.head()
```

lets convert integer into the string which they belong to ...


```python
def label_int2str(row):
    return emotions["train"].features["label"].int2str(row)
```

```python
df["label_name"] = df["label"].apply(label_int2str)
df.head()
```

lets lookintothe class distribution

```python
import matplotlib.pyplot as plt

df["label_name"].value_counts(ascending = True).plot.barh()
plt.title("frequency of classes")
plt.show()
```

tokenizing the datasets

```python
from transformers import AutoTokenizer
text = "Tokenizing text is a core task of NLP."

model_ckpt ="distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_ckpt)
encode_text = tokenizer(text)
print(encode_text)
```

```python
def tokenize(batch):
  return tokenizer(batch["text"], padding=True, truncation=True)
```

```python
tokenizer.vocab_size
```

```python
emotions.reset_format()
emotion_encode = emotions.map(tokenize, batched=True, batch_size=None)
print(tokenize(emotions["train"][:1]))
```
